```python
import numpy as np
```

```python
# Comparando lista Python normal com array Numpy

lista_python = [1, 2, 3, 4, 5]
array_numpy = np.array ([1, 2, 3, 4, 5])

print("Lista Python:", lista_python)
print("Array Numpy:", array_numpy)
print("Tipo do array:", type(array_numpy))
print("Versão do NumPy instalada:", np.__version__)

# Para multiplicar todos os elementos de uma lista por 2, seria necessário um loop: com NumPy pode ser feito diretamente:

print("Array multiplicado por 2:", array_numpy * 2)
```
