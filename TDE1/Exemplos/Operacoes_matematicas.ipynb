```python
import numpy as np
import time

# ---------------------------------------------------------
# 1) Operações aritméticas elemento a elemento
# ---------------------------------------------------------
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

soma        = a + b
subtracao   = b - a
multiplicacao = a * b
divisao     = b / a
potencia    = a ** 2

print("Soma:         ", soma)
print("Subtração:    ", subtracao)
print("Multiplicação:", multiplicacao)
print("Divisão:      ", divisao)
print("Potência:     ", potencia)

# ---------------------------------------------------------
# 2) Funções universais (ufuncs)
# ---------------------------------------------------------
c = np.array([1, 4, 9, 16, 25])

print("\nRaiz quadrada:", np.sqrt(c))
print("Exponencial:  ", np.exp(a))
print("Logaritmo:    ", np.log(c))
print("Seno:         ", np.sin(a))
print("Cosseno:      ", np.cos(a))

# ---------------------------------------------------------
# 3) Diferença de desempenho: ufuncs vs loop puro em Python
# ---------------------------------------------------------
n = 1_000_000
valores = np.arange(1, n + 1)

# Usando loop puro em Python
inicio = time.time()
resultado_loop = [np.sqrt(x) for x in valores]  # simula operação elemento a elemento
tempo_loop = time.time() - inicio

# Usando ufunc do NumPy (vetorizado, otimizado em C)
inicio = time.time()
resultado_ufunc = np.sqrt(valores)
tempo_ufunc = time.time() - inicio

print(f"\nTempo com loop Python puro: {tempo_loop:.4f} s")
print(f"Tempo com ufunc (np.sqrt):  {tempo_ufunc:.4f} s")
print(f"Ufunc foi aproximadamente {tempo_loop / tempo_ufunc:.1f}x mais rápido")

# ---------------------------------------------------------
# 4) Operações de agregação
# ---------------------------------------------------------
d = np.array([1, 2, 3, 4, 5])

print("\nSoma total (sum):       ", np.sum(d))
print("Produto total (prod):   ", np.prod(d))
print("Soma acumulada (cumsum):", np.cumsum(d))
```
