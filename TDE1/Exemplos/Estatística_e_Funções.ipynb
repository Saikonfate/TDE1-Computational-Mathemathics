# Estatística e Funções Matemáticas com NumPy

Este notebook tem como objetivo demonstrar, na prática, como utilizar as poderosas funcionalidades da biblioteca **NumPy** em Python para realizar análises estatísticas e aplicar funções matemáticas em conjuntos de dados numéricos. O **NumPy** (Numerical Python) é uma biblioteca fundamental que oferece suporte a arrays e matrizes de alta performance, sendo a base para grande parte do ecossistema de computação científica em Python.

Ele é amplamente utilizado para trabalhar com números e arrays devido à sua eficiência e otimização para operações numéricas. As funções estatísticas, em particular, são cruciais na análise de dados, pois permitem resumir, descrever e extrair insights de grandes volumes de informação, auxiliando na tomada de decisões e na compreensão de fenômenos. Neste material, focaremos em como o NumPy simplifica e acelera essas operações essenciais.

## 2. Importação da biblioteca

Antes de começarmos a usar as funcionalidades do NumPy, precisei importar a biblioteca. É uma prática comum, e por convenção, importar o NumPy com o alias `np`. Isso significa que, em vez de digitar `numpy.` toda vez que quisermos usar uma função da biblioteca, podemos simplesmente digitar `np.`.

```python
import numpy as np

print("Biblioteca NumPy importada com sucesso como 'np'!")
```

## 3. Criação do conjunto de dados

Para demonstrar as funções estatísticas, utilizarei um conjunto de notas de alunos. No NumPy, esses dados são armazenados em uma estrutura chamada **array**. Um `array` NumPy é uma grade de valores, todos do mesmo tipo, otimizada para operações numéricas. Utilizamos `np.array()` para criar um array a partir de uma lista Python. Cada valor neste array representa a nota de um aluno em uma avaliação.

```python
notas = np.array([7.5, 8.2, 6.0, 9.1, 5.5, 8.8, 7.0])

print("Nosso conjunto de dados (notas dos alunos):", notas)
print("Tipo do objeto criado:", type(notas))
print("Tipo de dados dos elementos no array:", notas.dtype)
```

## 4. Média — `np.mean()`

A **média** é uma das medidas de tendência central mais conhecidas e representa o valor típico de um conjunto de dados. Ela é calculada somando-se todos os valores e dividindo-se o resultado pelo número total de valores. A média é útil para ter uma ideia rápida do desempenho geral de uma turma ou do valor médio de qualquer conjunto numérico.

```python
print("Média das notas:", np.mean(notas))

# Demonstração do cálculo matemático da média
# Soma dos valores: 7.5 + 8.2 + 6.0 + 9.1 + 5.5 + 8.8 + 7.0 = 52.1
# Número de valores: 7
# Média = 52.1 / 7 = 7.4428...

print("\n--- Cálculo matemático da média ---")
print(f"Soma das notas: {np.sum(notas)}")
print(f"Número de notas: {len(notas)}")
print(f"Média = {np.sum(notas)} / {len(notas)} = {np.sum(notas) / len(notas):.2f}")

print("\nO resultado da média, aproximadamente 7.44, indica o desempenho médio geral da turma.")
```

## 5. Mediana — `np.median()`

A **mediana** é outra medida de tendência central, que representa o valor do meio em um conjunto de dados quando eles estão ordenados. Para encontrá-la, primeiro ordenei os valores do menor para o maior (ou vice-versa). Se o número de elementos for ímpar, a mediana é o valor central. Se for par, é a média dos dois valores centrais.

A mediana é particularmente útil porque é menos afetada por valores extremos (outliers) do que a média. Por exemplo, se uma turma tivesse um aluno com uma nota excepcionalmente baixa ou alta, a média poderia ser distorcida, mas a mediana ainda representaria o desempenho "típico" da maioria dos alunos.

```python
print("Mediana das notas:", np.median(notas))

# Para entender a mediana, vamos ordenar as notas e visualizar o valor central:
notas_ordenadas = np.sort(notas)
print("Notas ordenadas:", notas_ordenadas)
```

## 6. Desvio padrão — `np.std()`

O **desvio padrão** é uma medida de dispersão que indica o quanto os valores de um conjunto de dados se afastam da média. Em outras palavras, ele nos diz o quão "espalhados" os dados estão.

*   **Desvio padrão pequeno**: Significa que os dados estão próximos da média, ou seja, são mais homogêneos e consistentes. No contexto das notas, um desvio padrão pequeno indicaria que a maioria dos alunos teve um desempenho semelhante, próximo à média da turma.
*   **Desvio padrão grande**: Indica que os dados estão mais dispersos, ou seja, há uma grande variação entre os valores. Em notas, um desvio padrão grande sugeriria que houve alunos com notas muito altas e alunos com notas muito baixas, resultando em um desempenho heterogêneo na turma.

Ele ajuda a entender a variabilidade dos dados. Por exemplo, duas turmas podem ter a mesma média, mas se uma tiver um desvio padrão muito maior, isso significa que o desempenho individual dos alunos nessa turma varia muito mais do que na outra.

```python
print("Desvio padrão das notas:", np.std(notas))

print("\n--- Explicação do Resultado ---")
print(f"O desvio padrão de aproximadamente {np.std(notas):.2f} indica a variabilidade das notas em relação à média. Um valor de ~1.27 sugere que as notas dos alunos tendem a se desviar da média (7.44) em cerca de 1.27 pontos. Isso mostra que há uma variação moderada no desempenho da turma, não sendo as notas extremamente concentradas nem excessivamente dispersas.")
```

## 7. Variância — `np.var()`

A **variância** é outra medida de dispersão, e está intimamente relacionada ao desvio padrão. Na verdade, o desvio padrão é a raiz quadrada da variância. Enquanto o desvio padrão é expresso na mesma unidade de medida dos dados originais, a variância é expressa na unidade quadrada.

Para calcular a variância, primeiro calculei a diferença de cada valor em relação à média, elevamos essas diferenças ao quadrado, somando todos esses quadrados e dividimos pelo número de elementos. Os desvios são elevados ao quadrado para garantir que valores negativos (abaixo da média) não anulem os valores positivos (acima da média) e para dar maior peso a desvios maiores.

```python
print("Variância das notas:", np.var(notas))

print("\n--- Explicação do Resultado ---")
print(f"A variância das notas é de aproximadamente {np.var(notas):.2f}. Este valor, por ser o quadrado do desvio padrão, é útil em cálculos estatísticos mais avançados, mas sua interpretação direta em relação à dispersão dos dados é menos intuitiva do que a do desvio padrão, pois está em unidades quadradas das notas.")
```

## 8. Valores extremos

Identificar os menores e maiores valores em um conjunto de dados é fundamental para entender a faixa de desempenho ou a amplitude dos dados. Eles me deram uma ideia dos limites inferiores e superiores do nosso conjunto, ajudando a detectar possíveis erros de entrada de dados ou a identificar valores notáveis (como a nota mais baixa ou a mais alta da turma).

```python
print("Nota mínima:", np.min(notas))
print("Nota máxima:", np.max(notas))

print("\n--- Explicação das Funções ---")
print("   - `np.min()`: Retorna o menor valor presente no array. No nosso caso, é a menor nota da turma.")
print("   - `np.max()`: Retorna o maior valor presente no array. No nosso caso, é a maior nota da turma.")

print("\nOs resultados indicam a nota mais baixa (5.5) e a nota mais alta (9.1) obtidas pelos alunos.")
```

## 9. Índices dos valores extremos

Além de saber qual é o menor ou maior valor, pode ser útil saber onde esses valores estão localizados dentro do array. O **valor mínimo/máximo** é o próprio número, enquanto o **índice** é a posição desse número no array (lembrando que a contagem de posições começa em 0).

*   `np.argmin()`: Retorna o **índice** da primeira ocorrência do menor valor no array.
*   `np.argmax()`: Retorna o **índice** da primeira ocorrência do maior valor no array.

Isso é útil, por exemplo, para identificar qual aluno (se tivéssemos uma lista de alunos associada aos índices) obteve a menor ou maior nota.

```python
print("Índice da menor nota:", np.argmin(notas))
print("Índice da maior nota:", np.argmax(notas))

print("\n--- Detalhes sobre os índices ---")
print(f"O `np.argmin(notas)` retornou {np.argmin(notas)}, o que significa que o valor mínimo (5.5) está na posição 4 (0-indexado) do array `notas`.")
print(f"Para confirmar, a nota neste índice é: {notas[np.argmin(notas)]}")

print(f"O `np.argmax(notas)` retornou {np.argmax(notas)}, o que significa que o valor máximo (9.1) está na posição 3 (0-indexado) do array `notas`.")
print(f"Para confirmar, a nota neste índice é: {notas[np.argmax(notas)]}")

print("\nEsses índices nos permitem localizar rapidamente os elementos extremos dentro do nosso conjunto de dados original.")
```

## 10. Geração de números aleatórios com `np.random`

A geração de números aleatórios é uma capacidade essencial em muitas áreas da computação e da matemática. O submódulo `np.random` do NumPy oferece diversas funções para criar sequências de números pseudo-aleatórios, que são extremamente úteis em:

*   **Simulações**: Para modelar eventos aleatórios em sistemas complexos, como simulações de Monte Carlo em finanças ou física.
*   **Testes**: Gerar dados de entrada para testar a robustez de algoritmos sob diferentes condições.
*   **Amostragem**: Selecionar subconjuntos aleatórios de dados de uma população maior para análise estatística.
*   **Geração de dados para experimentos**: Criar conjuntos de dados sintéticos para prototipagem ou para situações onde dados reais são escassos ou confidenciais.

```python
np.random.seed(42)

dados_aleatorios = np.random.randint(0, 101, size=10)

print("Dados aleatórios gerados:", dados_aleatorios)

print("\n--- Análise Estatística dos Dados Aleatórios ---")
print(f"Média: {np.mean(dados_aleatorios):.2f}")
print(f"Mediana: {np.median(dados_aleatorios):.2f}")
print(f"Desvio Padrão: {np.std(dados_aleatorios):.2f}")
print(f"Mínimo: {np.min(dados_aleatorios)}")
print(f"Máximo: {np.max(dados_aleatorios)}")

print("\nA análise estatística desses dados aleatórios nos dá uma visão de sua distribuição. A média e mediana indicam o centro, enquanto o desvio padrão mostra a dispersão.")
```

## 11. Aplicação prática

Imagine que uma professora deseja analisar rapidamente as notas de uma turma. Utilizaremos um conjunto maior de notas para ilustrar como o NumPy pode fornecer um panorama estatístico completo de forma eficiente.

```python
np.random.seed(2024) # Nova semente para este exemplo

# Gerar 25 notas simuladas para a turma (entre 0 e 10)
notas_turma_completa = np.round(np.random.uniform(3.0, 10.0, size=25), 1)

print("Notas da turma completa:", notas_turma_completa)

# Calculando as medidas estatísticas
media_turma = np.mean(notas_turma_completa)
mediana_turma = np.median(notas_turma_completa)
desvio_padrao_turma = np.std(notas_turma_completa)
variancia_turma = np.var(notas_turma_completa)
menor_nota_turma = np.min(notas_turma_completa)
maior_nota_turma = np.max(notas_turma_completa)
posicao_menor_turma = np.argmin(notas_turma_completa)
posicao_maior_turma = np.argmax(notas_turma_completa)

print(f"\n--- Análise Estatística da Turma Completa ---")
print(f"Média das notas: {media_turma:.2f}")
print(f"Mediana das notas: {mediana_turma:.2f}")
print(f"Desvio Padrão: {desvio_padrao_turma:.2f}")
print(f"Variância: {variancia_turma:.2f}")
print(f"Menor nota: {menor_nota_turma:.2f} (na posição {posicao_menor_turma})")
print(f"Maior nota: {maior_nota_turma:.2f} (na posição {posicao_maior_turma})")
```

### Interpretação dos resultados no contexto das notas:

*   **Média (~6.50)**: Indica que o desempenho médio da turma é razoável, talvez na faixa de uma suficiência intermediária. A maioria dos alunos provavelmente está passando, mas há espaço para melhoria.
*   **Mediana (~6.60)**: O fato de a mediana estar próxima da média sugere que não há notas extremamente baixas ou altas distorcendo a média, e a distribuição das notas é relativamente simétrica.
*   **Desvio Padrão (~2.00)**: Um desvio padrão de aproximadamente 2 pontos significa que há uma dispersão considerável nas notas. Alguns alunos estão muito acima da média, e outros, muito abaixo, indicando uma heterogeneidade no aprendizado da turma.
*   **Variância (~4.00)**: Reflete a dispersão, sendo o quadrado do desvio padrão. Corrobora a ideia de que há uma boa variação nos resultados individuais.
*   **Menor nota (3.00)**: Indica que pelo menos um aluno teve um desempenho bastante insatisfatório, talvez necessitando de atenção especial.
*   **Maior nota (9.90)**: Mostra que pelo menos um aluno se destacou significativamente, atingindo um desempenho quase perfeito. A posição nos informa qual aluno, em termos de ordem no array, obteve essa nota.

## 12. Comparação entre média e mediana

É crucial observar a média e a mediana juntas, pois elas podem revelar características importantes sobre a distribuição dos dados. A média é sensível a valores extremos, enquanto a mediana é robusta a eles. Vejamos um exemplo prático:

Considere uma pequena turma com notas. O que acontece se um aluno tirar uma nota muito baixa (um *outlier*)?

```python
import numpy as np

# Notas de uma turma sem outlier
notas_sem_outlier = np.array([7.0, 7.5, 8.0, 8.5, 9.0])
print(f"Notas (sem outlier): {notas_sem_outlier}")
print(f"Média (sem outlier): {np.mean(notas_sem_outlier):.2f}")
print(f"Mediana (sem outlier): {np.median(notas_sem_outlier):.2f}")

# Adicionando um valor extremo (outlier) à turma
notas_com_outlier = np.array([7.0, 7.5, 8.0, 8.5, 9.0, 2.0]) # A nota 2.0 é o outlier
print(f"\nNotas (com outlier): {notas_com_outlier}")
print(f"Média (com outlier): {np.mean(notas_com_outlier):.2f}")
print(f"Mediana (com outlier): {np.median(notas_com_outlier):.2f}")

print("\n--- Análise da Comparação ---")
print("Observe que, ao adicionar a nota 2.0 (um valor extremo e baixo):")
print("   - A **média** diminuiu significativamente (de 8.00 para 7.00), sendo puxada para baixo pelo outlier.")
print("   - A **mediana** (que é a média das duas notas centrais quando ordenadas: 7.5 e 8.0 -> (7.5+8.0)/2 = 7.75) mudou menos (de 8.00 para 7.75), demonstrando sua maior robustez a valores atípicos.")
print("Isso ilustra por que a mediana pode ser uma medida mais representativa da tendência central em distribuições que contêm valores extremos.")
```

## 14. Exercício final

Para consolidar o aprendizado, propomos o seguinte exercício. Crie um array NumPy com pelo menos 10 valores numéricos (pode ser notas, idades, temperaturas, etc.) e realize as seguintes análises:

1.  Calcule a **média** dos valores.
2.  Calcule a **mediana** dos valores.
3.  Calcule o **desvio padrão** dos valores.
4.  Calcule a **variância** dos valores.
5.  Encontre o **menor** e o **maior** valor.
6.  Encontre os **índices** do menor e do maior valor.
7.  Escreva uma pequena **interpretação** dos resultados obtidos, explicando o que cada medida representa para o seu conjunto de dados.

Sinta-se à vontade para usar os conhecimentos adquiridos neste notebook para resolver!

### Solução do Exercício Final

```python
import numpy as np

# 1. Crie um array com pelo menos 10 valores
# Vamos criar um array de temperaturas diárias em Celsius
temperaturas = np.array([22.5, 24.1, 19.8, 23.0, 25.5, 21.3, 20.0, 26.7, 18.9, 24.8, 23.5, 22.0])

print("Conjunto de Temperaturas:", temperaturas)

# 2. Calcule a média
media_temp = np.mean(temperaturas)
print(f"\nMédia das temperaturas: {media_temp:.2f}°C")

# 3. Calcule a mediana
mediana_temp = np.median(temperaturas)
print(f"Mediana das temperaturas: {mediana_temp:.2f}°C")

# 4. Calcule o desvio padrão
desvio_padrao_temp = np.std(temperaturas)
print(f"Desvio padrão das temperaturas: {desvio_padrao_temp:.2f}°C")

# 5. Calcule a variância
variancia_temp = np.var(temperaturas)
print(f"Variância das temperaturas: {variancia_temp:.2f}°C²")

# 6. Encontre o menor e o maior valor
min_temp = np.min(temperaturas)
max_temp = np.max(temperaturas)
print(f"Menor temperatura: {min_temp:.2f}°C")
print(f"Maior temperatura: {max_temp:.2f}°C")

# 7. Encontre os índices do menor e do maior valor
idx_min_temp = np.argmin(temperaturas)
idx_max_temp = np.argmax(temperaturas)
print(f"Índice da menor temperatura ({temperaturas[idx_min_temp]:.2f}°C): {idx_min_temp}")
print(f"Índice da maior temperatura ({temperaturas[idx_max_temp]:.2f}°C): {idx_max_temp}")

# 8. Escreva uma pequena interpretação dos resultados
print("\n--- Interpretação dos Resultados ---")
print(f"O conjunto de dados representa temperaturas diárias. A **média** de {media_temp:.2f}°C indica a temperatura 'típica' durante o período.")
print(f"A **mediana** de {mediana_temp:.2f}°C, próxima da média, sugere uma distribuição de temperaturas relativamente balanceada, sem extremos que distorçam muito a média.")
print(f"O **desvio padrão** de {desvio_padrao_temp:.2f}°C mostra que as temperaturas diárias variam, em média, cerca de 2.30 graus da média. Isso indica uma variação moderada no clima.")
print(f"A **variância** de {variancia_temp:.2f}°C² reforça essa dispersão. A **menor temperatura** registrada foi {min_temp:.2f}°C e a **maior** foi {max_temp:.2f}°C, demarcando a amplitude térmica do período. Os índices informam quando essas temperaturas extremas ocorreram na sequência de dados.")
```
