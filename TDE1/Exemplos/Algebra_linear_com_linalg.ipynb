```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
B = np.array([[1, 0], [0, 1]])

print("Multiplicação de matrizes (A @ B):\n", A @ B)
print("Transposta de A:\n", A.T)
print("Determinante de A:", np.linalg.det(A))
print("Inversa de A:\n", np.linalg.inv(A))

# Resolvendo o sistema linear: 2x + y = 5 ; x + 3y = 10
coeficientes = np.array([[2, 1], [1, 3]])
resultados = np.array([5, 10])
solucao = np.linalg.solve(coeficientes, resultados)
print("Solução do sistema (x, y):", solucao)

autovalores, autovetores = np.linalg.eig(A)
print("Autovalores:", autovalores)
```
