```python
import numpy as np
import time

# Demonstrando a limitação de tipo único e a vantagem de desempenho
tamanho = 2_000_000
lista = list(range(tamanho))
array = np.arange(tamanho)

inicio = time.time()
soma_lista = sum([x * 2 for x in lista])
tempo_lista = time.time() - inicio

inicio = time.time()
soma_array = np.sum(array * 2)
tempo_array = time.time() - inicio

print(f"Soma com lista Python: {soma_lista} | tempo: {tempo_lista:.4f}s")
print(f"Soma com array NumPy: {soma_array} | tempo: {tempo_array:.4f}s")
print(f"NumPy foi aproximadamente {tempo_lista/tempo_array:.1f}x mais rápido")

# Limitação: tipo único no array
array_misto = np.array([1, "texto", 3.5])
print("Ao misturar tipos, o NumPy converte tudo para string:", array_misto, array_misto.dtype)
```
