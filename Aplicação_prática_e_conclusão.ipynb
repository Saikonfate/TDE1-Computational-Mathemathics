````python
```python
import numpy as np
import matplotlib.pyplot as plt

# Simulando uma "imagem" em escala de cinza como matriz NumPy
imagem = np.random.randint(0, 256, size=(10, 10))
print("Matriz da imagem (10x10):\n", imagem)

# Aplicação matemática: normalizar os valores entre 0 e 1
imagem_normalizada = imagem / 255.0
print("\nImagem normalizada (0 a 1):\n", imagem_normalizada.round(2))

# Aplicação prática: inverter as cores (negativo da imagem)
imagem_invertida = 255 - imagem

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1)
plt.title("Imagem original")
plt.imshow(imagem, cmap="gray")
plt.subplot(1, 2, 2)
plt.title("Imagem invertida")
plt.imshow(imagem_invertida, cmap="gray")
plt.show()
```
````
